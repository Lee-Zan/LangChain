```python
# 2. Dropdown Template


from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key="GOOGLE_API_KEY"
)

def choose(title, options):
    print(f"\n{title}")
    for i, option in enumerate(options, 1):
        print(f"{i}. {option}")
    return options[int(input("Choose: ")) - 1]

paper = choose("Research Paper", [
    "Attention Is All You Need",
    "BERT",
    "GPT-3",
    "Diffusion Models"
])

style = choose("Style", [
    "Beginner-Friendly",
    "Technical",
    "Code-Oriented",
    "Mathematical"
])

length = choose("Length", [
    "Short",
    "Medium",
    "Long"
])

prompt = f"Explain '{paper}' in {style} style with {length} detail."

print("\nGenerating...\n")
print(model.invoke(prompt).content)
```

```python
# 3. Template


from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate

GOOGLE_API_KEY = "GOOGLE_API_KEY"

model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=GOOGLE_API_KEY
)


papers = [
    "Attention Is All You Need",
    "BERT: Pre-training of Deep Bidirectional Transformers",
    "GPT-3: Language Models are Few-Shot Learners",
    "Diffusion Models Beat GANs on Image Synthesis"
]

styles = [
    "Beginner-Friendly",
    "Technical",
    "Code-Oriented",
    "Mathematical"
]

lengths = [
    "Short (1-2 paragraphs)",
    "Medium (3-5 paragraphs)",
    "Long (Detailed explanation)"
]

print("Select Research Paper:")
for i, p in enumerate(papers, 1):
    print(f"{i}. {p}")
paper_input = papers[int(input("Choose: ")) - 1]

print("\nSelect Style:")
for i, s in enumerate(styles, 1):
    print(f"{i}. {s}")
style_input = styles[int(input("Choose: ")) - 1]

print("\nSelect Length:")
for i, l in enumerate(lengths, 1):
    print(f"{i}. {l}")
length_input = lengths[int(input("Choose: ")) - 1]


template = PromptTemplate(
    template="""
Please summarize the research paper titled "{paper_input}"

Explanation Style: {style_input}
Explanation Length: {length_input}

1. Mathematical Details:
- Include relevant mathematical equations if present.
- Explain mathematical concepts simply.

2. Analogies:
- Use relatable analogies to simplify complex ideas.

If information is unavailable, respond with:
"Insufficient information available"

Ensure the summary matches the requested style and length.
""",
    input_variables=["paper_input", "style_input", "length_input"]
)


prompt = template.invoke({
    "paper_input": paper_input,
    "style_input": style_input,
    "length_input": length_input
})


result = model.invoke(prompt)

print("\n" + "="*60)
print("SUMMARY")
print("="*60)
print(result.content)
```

```python
# 4. Chatbot



from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage

GOOGLE_API_KEY = "GOOGLE_API_KEY"

model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=GOOGLE_API_KEY
)

chat_history = [
    SystemMessage(content="You are a helpful AI assistant")
]

while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        break

    chat_history.append(HumanMessage(content=user_input))

    result = model.invoke(chat_history)

    ai_text = str(result.content)

    print("AI:", ai_text)

    chat_history.append(AIMessage(content=ai_text))

print("\n--- Chat History ---")

for msg in chat_history:
    print(f"{msg.type}: {msg.content}")
```

```python
# 5. Message




from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
from langchain_google_genai import ChatGoogleGenerativeAI

GOOGLE_API_KEY = "YOUR_GEMINI_API_KEY"


model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=GOOGLE_API_KEY
)


messages = [
    SystemMessage(content="You are a helpful assistant"),
    HumanMessage(content="Tell me about LangChain")
]

result = model.invoke(messages)

ai_text = str(result.content)

messages.append(AIMessage(content=ai_text))

for msg in messages:
    print(f"{msg.type}: {msg.content}\n")
```

```python
# 6. Chat Prompt Template



from langchain_core.prompts import ChatPromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI

GOOGLE_API_KEY = "YOUR_GEMINI_API_KEY"

model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=GOOGLE_API_KEY
)

chat_template = ChatPromptTemplate([
    ("system", "You are a helpful {role} expert"),
    ("human", "Explain in easy and simple words, what is {topic}")
])

prompt = chat_template.invoke({
    "role": "football",
    "topic": "offside trap"
})

result = model.invoke(prompt)

print(result.content)
```

```python
# 7. Message Placeholder



from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import HumanMessage, AIMessage

chat_template = ChatPromptTemplate([
    ("system", "You are a helpful customer support agent."),
    MessagesPlaceholder(variable_name="chat_history"),
    ("human", "{query}")
])

chat_history = [
    HumanMessage(content="I ordered a laptop last week."),
    AIMessage(content="Sure. Could you provide your order number?"),
    HumanMessage(content="My order number is 12345.")
]

prompt = chat_template.invoke({
    "chat_history": chat_history,
    "query": "Where is my refund?"
})

for msg in prompt.messages:
    print(f"{msg.type}: {msg.content}")
```
