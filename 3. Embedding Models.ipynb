```python
# OpenAI


from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv

load_dotenv()

embedding = OpenAIEmbeddings(model = 'text-embedding-3-large', dimensions = 32)

result = embedding.embed_query("Kathmandu is the capital city of nepal.")

print(str(result))
```

```python
# 2. OpenAI docs


from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv

load_dotenv()

embedding = OpenAIEmbeddings(model = 'text-embedding-3-large', dimensions = 32)

documents = [
    "Kathmandu is the capital city of Nepal.",
    "New Delhi is the capital city of India.",
    "Beijing is the capital city of China."
]

result = embedding.embed_documents(documents)

print(str(result))
```

```python
# 3. HuggingFace local


from langchain_huggingface import HuggingFaceEmbeddings

embedding = HuggingFaceEmbeddings(model_name='sentence-transformers/all-MiniLM-L6-v2')

text = "Kathmandu is the capital city of Nepal."

vector = embedding.embed_query(text)

print(str(vector))
```

```python
# 4. Document Similarity



from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv

from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

embedding = OpenAIEmbeddings(model='text-embedding-3-large', dimensions=300)

documents = [
    "CSK - Chennai Super Kings: Dhoni’s legacy team, calm and consistent under Ruturaj Gaikwad.",
    "MI - Mumbai Indians: Five-time champions, built on Rohit Sharma’s batting and Bumrah’s bowling.",
    "RCB - Royal Challengers Bengaluru: Defending champions led by Virat Kohli, chasing sustained dominance.",
    "KKR - Kolkata Knight Riders: Two-time winners, known for flair and unpredictable brilliance.",
    "SRH - Sunrisers Hyderabad: Strong bowling unit, now led by Pat Cummins for stability.",
    "PBKS - Punjab Kings: Energetic side with Shreyas Iyer, still hunting their first IPL title.",
    "GT - Gujarat Titans: Young franchise with Shubman Gill and Rashid Khan, already proven winners.",
    "DC - Delhi Capitals: Talented squad under Axar Patel, aiming to break their title drought.",
    "RR - Rajasthan Royals: First-ever IPL champions, banking on Yashasvi Jaiswal and Riyan Parag.",
    "LSG - Lucknow Super Giants: New-age team with Rishabh Pant and Nicholas Pooran, hungry for silverware."
]

query = 'Tell me about MI'

doc_embeddings = embedding.embed_documents(documents)
query_embedding = embedding.embed_query(query)

scores = cosine_similarity([query_embedding], doc_embeddings)[0]

index, score = sorted(list(enumerate(scores)),key=lambda x:x[1])[-1]

print(query)

print(documents[index])

print("Similarity score is",score)
```
