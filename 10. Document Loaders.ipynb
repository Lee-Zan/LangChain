```python
# 1. TextLoader




from google.colab import files
from langchain_community.document_loaders import TextLoader
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate

GOOGLE_API_KEY = "GOOGLE_API_KEY"

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=GOOGLE_API_KEY
)

prompt = PromptTemplate(
    template="""
Write a summary for the following poem:

{poem}
""",
    input_variables=["poem"]
)

parser = StrOutputParser()

loader = TextLoader(file_name)

docs = loader.load()

print("Document Preview:\n")
print(docs[0].page_content[:500])

chain = prompt | model | parser

result = chain.invoke(
    {
        "poem": docs
    }
)

print("\nSummary:\n")
print(result)
```

```python
# 2. PDF Loader using PyPDFLoader



from google.colab import files
from langchain_community.document_loaders import PyPDFLoader

uploaded = files.upload()

pdf_file = list(uploaded.keys())[0]

loader = PyPDFLoader(pdf_file)

docs = loader.load()

print("First Page Content:\n")
print(docs[0].page_content)

print("\n" + "=" * 80 + "\n")

print("Second Page Metadata:\n")
print(docs[1].metadata)
```

```python
# Part 4 : View Specific Page Content and Metadata


print("Page Content:\n")
print(docs[1].page_content)

print("\nMetadata:\n")
print(docs[1].metadata)
```

```python
# Part 5 : Display All Loaded Pages


for i, doc in enumerate(docs):
    print("\n" + "=" * 80)
    print(f"Page {i+1}")
    print("=" * 80)

    print(doc.page_content[:1000])

    print("\nMetadata:")
    print(doc.metadata)
```

```python
# 4. WebBaseLoader (Interactive Version)



from langchain_community.document_loaders import WebBaseLoader
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate

GOOGLE_API_KEY = "GOOGLE_API_KEY"

url = input("Enter URL: ")

loader = WebBaseLoader(url)

docs = loader.load()

model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=GOOGLE_API_KEY
)

prompt = PromptTemplate(
    template="""
Answer the following question:

Question:
{question}

Webpage Content:
{text}
""",
    input_variables=["question", "text"]
)

parser = StrOutputParser()

chain = prompt | model | parser

question = input("Ask a question about the webpage: ")

result = chain.invoke(
    {
        "question": question,
        "text": docs[0].page_content
    }
)

print("\nAnswer:")
print(result)
```

```python
# 5. CSVLoader




from google.colab import files
from langchain_community.document_loaders import CSVLoader

uploaded = files.upload()

csv_file = list(uploaded.keys())[0]

loader = CSVLoader(file_path=csv_file)

docs = loader.load()

print("Total Rows Loaded:", len(docs))

while True:
    row = input(f"\nEnter row number (0 to {len(docs)-1}) or 'exit': ")

    if row.lower() == "exit":
        break

    row = int(row)

    if 0 <= row < len(docs):
        print("\nRow Content:")
        print(docs[row].page_content)

        print("\nMetadata:")
        print(docs[row].metadata)
    else:
        print("Invalid row number.")
```

```python
# 6. Custom Document Loader (fixed task as defined in the code)




from google.colab import files
from langchain_community.document_loaders import TextLoader, PyPDFLoader, CSVLoader, DirectoryLoader, WebBaseLoader
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
import os
import shutil

GOOGLE_API_KEY = "GOOGLE_API_KEY"

model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=GOOGLE_API_KEY
)

parser = StrOutputParser()

class CustomLoader:

    def load_document(self, doc_type):

        if doc_type == "txt":

            uploaded = files.upload()
            file_name = list(uploaded.keys())[0]

            docs = TextLoader(file_name).load()

            prompt = PromptTemplate(
                template="Write a summary of the following text:\n{content}",
                input_variables=["content"]
            )

            chain = prompt | model | parser

            print(chain.invoke({
                "content": docs[0].page_content
            }))

        elif doc_type == "pdf":

            uploaded = files.upload()
            file_name = list(uploaded.keys())[0]

            docs = PyPDFLoader(file_name).load()

            full_text = "\n".join(
                doc.page_content for doc in docs
            )

            prompt = PromptTemplate(
                template="Summarize the following PDF:\n{content}",
                input_variables=["content"]
            )

            chain = prompt | model | parser

            print(chain.invoke({
                "content": full_text
            }))

        elif doc_type == "csv":

            uploaded = files.upload()
            file_name = list(uploaded.keys())[0]

            docs = CSVLoader(
                file_path=file_name
            ).load()

            data = "\n".join(
                doc.page_content for doc in docs[:10]
            )

            prompt = PromptTemplate(
                template="Analyze the following CSV data:\n{content}",
                input_variables=["content"]
            )

            chain = prompt | model | parser

            print(chain.invoke({
                "content": data
            }))

        elif doc_type == "directory":

            uploaded = files.upload()

            os.makedirs("Notes", exist_ok=True)

            for file_name in uploaded.keys():
                shutil.move(
                    file_name,
                    f"Notes/{file_name}"
                )

            docs = DirectoryLoader(
                path="Notes",
                glob="*.pdf",
                loader_cls=PyPDFLoader
            ).load()

            text = "\n".join(
                doc.page_content for doc in docs[:5]
            )

            prompt = PromptTemplate(
                template="Summarize these documents:\n{content}",
                input_variables=["content"]
            )

            chain = prompt | model | parser

            print(chain.invoke({
                "content": text
            }))

        elif doc_type == "web":

            url = input("Enter URL: ")

            docs = WebBaseLoader(url).load()

            question = input(
                "Enter Your Question: "
            )

            prompt = PromptTemplate(
                template="""
Answer the following question.

Question:
{question}

Text:
{content}
""",
                input_variables=[
                    "question",
                    "content"
                ]
            )

            chain = prompt | model | parser

            print(chain.invoke({
                "question": question,
                "content": docs[0].page_content
            }))

        else:
            print("Invalid Document Type")

loader = CustomLoader()

print("Available Loaders")
print("1. txt")
print("2. pdf")
print("3. csv")
print("4. directory")
print("5. web")

choice = input(
    "\nWhich document would you like to load? "
).lower()

loader.load_document(choice)
```

```python
# 6. Custom Document Loader (Custom task as defined by the user)




from google.colab import files
from langchain_community.document_loaders import TextLoader, PyPDFLoader, CSVLoader, DirectoryLoader, WebBaseLoader
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
import os
import shutil

GOOGLE_API_KEY = "GOOGLE_API_KEY"

model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=GOOGLE_API_KEY
)

parser = StrOutputParser()

prompt = PromptTemplate(
    template="""
You are a helpful assistant.

Answer the following question using only the provided content.

Question: {question}
Content: {content}

""",
    input_variables=["question", "content"]
)

chain = prompt | model | parser


class CustomLoader:

    def ask_question(self, content):
        question = input("\nEnter your question: ")

        result = chain.invoke(
            {
                "question": question,
                "content": content
            }
        )

        print("\nAnswer:\n")
        print(result)

    def load_document(self, doc_type):

        if doc_type == "txt":

            uploaded = files.upload()

            file_name = list(uploaded.keys())[0]

            docs = TextLoader(file_name).load()

            self.ask_question(
                docs[0].page_content
            )

        elif doc_type == "pdf":

            uploaded = files.upload()

            file_name = list(uploaded.keys())[0]

            docs = PyPDFLoader(file_name).load()

            content = "\n".join(
                doc.page_content
                for doc in docs
            )

            self.ask_question(content)

        elif doc_type == "csv":

            uploaded = files.upload()

            file_name = list(uploaded.keys())[0]

            docs = CSVLoader(
                file_path=file_name
            ).load()

            content = "\n".join(
                doc.page_content
                for doc in docs
            )

            self.ask_question(content)

        elif doc_type == "directory":

            uploaded = files.upload()

            os.makedirs(
                "Notes",
                exist_ok=True
            )

            for file_name in uploaded.keys():

                shutil.move(
                    file_name,
                    f"Notes/{file_name}"
                )

            docs = DirectoryLoader(
                path="Notes",
                glob="*.pdf",
                loader_cls=PyPDFLoader
            ).load()

            content = "\n".join(
                doc.page_content
                for doc in docs
            )

            self.ask_question(content)

        elif doc_type == "web":

            url = input(
                "Enter Website URL: "
            )

            docs = WebBaseLoader(
                url
            ).load()

            self.ask_question(
                docs[0].page_content
            )

        else:
            print(
                "Invalid Document Type"
            )


loader = CustomLoader()

print("\nAvailable Loaders")
print("1. txt")
print("2. pdf")
print("3. csv")
print("4. directory")
print("5. web")

choice = input(
    "\nWhich document would you like to load? "
).lower()

loader.load_document(choice)
```
