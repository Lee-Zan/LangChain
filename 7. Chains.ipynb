```python
# 1. Simple Chain



from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

GOOGLE_API_KEY = "GOOGLE_API_KEY"

prompt = PromptTemplate(
    template="Generate 5 very-short interesting facts about {topic}",
    input_variables=["topic"]
)

model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=GOOGLE_API_KEY
)

parser = StrOutputParser()

chain = prompt | model | parser

result = chain.invoke({"topic": "Nepal"})

print(result)
```

```python
# 2. Sequential Chain



from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

GOOGLE_API_KEY = "GOOGLE_API_KEY"

model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=GOOGLE_API_KEY
)

parser = StrOutputParser()

report_prompt = PromptTemplate(
    template="Generate a detailed report about {topic}",
    input_variables=["topic"]
)

summary_prompt = PromptTemplate(
    template="Create a concise 5-point summary of the following report:\n{report}",
    input_variables=["report"]
)

report_chain = report_prompt | model | parser

summary_chain = summary_prompt | model | parser

report = report_chain.invoke({
    "topic": "Nepal Premier League"
})

summary = summary_chain.invoke({
    "report": report
})

print("REPORT:\n")
print(report)

print("\n" + "=" * 60)

print("\nSUMMARY:\n")
print(summary)
```

```python
# Using RunnableSequence (Sequential Chain)



from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableSequence

GOOGLE_API_KEY = "GOOGLE_API_KEY"

model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=GOOGLE_API_KEY
)

parser = StrOutputParser()

chain = RunnableSequence(
    PromptTemplate.from_template(
        "Generate a detailed report about {topic}"
    ),
    model,
    parser,
    PromptTemplate.from_template(
        "Create a concise 5-point summary of the following report:\n{report}"
    ),
    model,
    parser
)

result = chain.invoke({
    "topic": "Nepal Football League"
})

print(result)
```

```python
# 3. Parallel Chain



from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableParallel

GOOGLE_API_KEY = "GEMINI_API_KEY"

model1 = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=GOOGLE_API_KEY
)

model2 = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=GOOGLE_API_KEY
)

prompt1 = PromptTemplate(
    template="Generate very short and simple notes from the following text:\n{text}",
    input_variables=["text"]
)

prompt2 = PromptTemplate(
    template="Generate 5 very short quiz questions from the following text:\n{text}",
    input_variables=["text"]
)

prompt3 = PromptTemplate(
    template="""
Merge the provided notes and quiz into a single document.

Notes:
{notes}

Quiz:
{quiz}
""",
    input_variables=["notes", "quiz"]
)

parser = StrOutputParser()

parallel_chain = RunnableParallel(
    {
        "notes": prompt1 | model1 | parser,
        "quiz": prompt2 | model2 | parser
    }
)

merge_chain = prompt3 | model1 | parser

chain = parallel_chain | merge_chain

text = """
Artificial Intelligence (AI) has become one of the most transformative technologies of the 21st century.
AI refers to computer systems designed to perform tasks that typically require human intelligence, such as learning, reasoning, problem-solving, decision-making, and natural language understanding.

The history of AI dates back to the 1950s when researchers began exploring whether machines could imitate human thinking. Early AI systems relied heavily on rule-based programming, where developers explicitly defined the logic computers should follow. While these systems worked well for specific tasks, they lacked flexibility and adaptability.

The rise of machine learning revolutionized artificial intelligence. Instead of being programmed with fixed rules, machine learning models learn patterns from data. These models can improve their performance over time as they are exposed to more information. Machine learning powers applications such as spam filters, recommendation systems, fraud detection, and predictive analytics.

Deep learning, a subset of machine learning, uses artificial neural networks inspired by the structure of the human brain. Deep learning has dramatically improved computer vision, speech recognition, and natural language processing. Technologies such as facial recognition, self-driving cars, and voice assistants rely heavily on deep learning algorithms.

One of the most significant developments in recent years has been the emergence of large language models (LLMs). These models are trained on vast amounts of text data and can generate human-like responses, summarize information, answer questions, write code, and perform a variety of language-related tasks. Popular examples include ChatGPT, Gemini, Claude, and other generative AI systems.

AI has had a major impact on healthcare. Medical professionals use AI-powered tools to analyze medical images, detect diseases, assist in diagnosis, and predict patient outcomes. AI can process large amounts of medical data much faster than humans, helping healthcare providers make informed decisions. Researchers are also exploring AI applications in drug discovery and personalized medicine.

In education, AI enables personalized learning experiences. Intelligent tutoring systems can adapt lessons based on a student's progress, strengths, and weaknesses. Automated grading, content recommendation, and virtual teaching assistants are also becoming increasingly common in educational environments.

The business sector has widely adopted AI to improve productivity and efficiency. Organizations use AI for customer support, supply chain optimization, demand forecasting, financial analysis, and marketing automation. Chatbots and virtual assistants help businesses provide 24/7 customer service while reducing operational costs.

AI is also transforming transportation. Autonomous vehicles use computer vision, sensors, and machine learning algorithms to understand their surroundings and make driving decisions. Although fully self-driving cars are still under development, AI-assisted driving systems are already available in many modern vehicles.

The entertainment industry uses AI to recommend movies, music, and other content based on user preferences. Streaming platforms analyze user behavior to provide personalized recommendations. AI is also used in game development, visual effects, content generation, and audience analytics.

Despite its benefits, AI presents several challenges and ethical concerns. Issues such as bias in algorithms, privacy risks, misinformation, job displacement, and the misuse of AI technologies require careful consideration. Governments, researchers, and technology companies are working to establish guidelines and regulations that promote responsible AI development.

The future of AI is expected to bring even greater advancements. Researchers are exploring more powerful models, multi-modal systems capable of understanding text, images, audio, and video simultaneously, and AI agents that can perform complex tasks autonomously. As AI continues to evolve, it will likely play an increasingly important role in shaping society, the economy, and everyday life.

In conclusion, artificial intelligence is a rapidly evolving field that has already transformed numerous industries and aspects of daily life. While challenges remain, the potential benefits of AI are immense. Success in the AI era will depend on balancing innovation with ethical responsibility, ensuring that these technologies are developed and used in ways that benefit humanity as a whole.
"""

result = chain.invoke({"text": text})

print(result)
```

```python
# Conditional Chain



from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser, PydanticOutputParser
from langchain_core.runnables import RunnableBranch, RunnableLambda
from pydantic import BaseModel, Field
from typing import Literal

GOOGLE_API_KEY = "GEMINI_API_KEY"

model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=GOOGLE_API_KEY
)

class Feedback(BaseModel):
    sentiment: Literal["Positive", "Negative"] = Field(
        description="Give the sentiment of the feedback"
    )

parser = StrOutputParser()
parser2 = PydanticOutputParser(pydantic_object=Feedback)

prompt1 = PromptTemplate(
    template="""
Classify the sentiment of the following feedback as Positive or Negative.

Feedback:
{feedback}

{format_instructions}
""",
    input_variables=["feedback"],
    partial_variables={
        "format_instructions": parser2.get_format_instructions()
    }
)

classifier_chain = prompt1 | model | parser2

prompt2 = PromptTemplate(
    template="""
Write a short 3-4 line response to this positive feedback:

{feedback}
""",
    input_variables=["feedback"]
)

prompt3 = PromptTemplate(
    template="""
Write a short 3-4 line response to this negative feedback:

{feedback}
""",
    input_variables=["feedback"]
)

branch_chain = RunnableBranch(
    (
        lambda x: x.sentiment == "Positive",
        RunnableLambda(
            lambda x: {"feedback": "Thank you for your positive feedback."}
        )
        | prompt2
        | model
        | parser
    ),
    (
        lambda x: x.sentiment == "Negative",
        RunnableLambda(
            lambda x: {"feedback": "We are sorry to hear about your experience."}
        )
        | prompt3
        | model
        | parser
    ),
    RunnableLambda(lambda x: "Could not determine sentiment.")
)

chain = classifier_chain | branch_chain

result = chain.invoke({
    "feedback": "This is a wonderful laptop. The battery life is amazing and the performance is excellent."
})

print(result)
```
