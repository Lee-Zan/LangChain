```python
# 1. Simple Chain



from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

GOOGLE_API_KEY = "GOOGLE_API_KEY"

prompt = PromptTemplate(
    template="Generate 5 very-short interesting facts about {topic}",
    input_variables=["topic"]
)

model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=GOOGLE_API_KEY
)

parser = StrOutputParser()

chain = prompt | model | parser

result = chain.invoke({"topic": "Nepal"})

print(result)
```

```python
# 2. Sequential Chain



from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

GOOGLE_API_KEY = "GOOGLE_API_KEY"

model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=GOOGLE_API_KEY
)

parser = StrOutputParser()

report_prompt = PromptTemplate(
    template="Generate a detailed report about {topic}",
    input_variables=["topic"]
)

summary_prompt = PromptTemplate(
    template="Create a concise 5-point summary of the following report:\n{report}",
    input_variables=["report"]
)

report_chain = report_prompt | model | parser

summary_chain = summary_prompt | model | parser

report = report_chain.invoke({
    "topic": "Nepal Premier League"
})

summary = summary_chain.invoke({
    "report": report
})

print("REPORT:\n")
print(report)

print("\n" + "=" * 60)

print("\nSUMMARY:\n")
print(summary)
```
